```python
import numpy as np
import scipy as sci
from scipy import stats
from scipy.stats import norm
import matplotlib.pyplot as plt
```

```python
mu, sigma = 0,2 
x_min=-6
x_max=6 
shag=0.1 
x=np.arange(x_min, x_max, shag) 
#Метод pdf(x, mu, sigma) объекта norm из библиотеки SciPy (модуль scipy.stats) вычисляет значение
#функции плотности вероятности (Probability Density Function, PDF) нормального распределения
#c параметрами mu и sigma в точках массива x .
y=norm.pdf(x,mu,sigma)
plt.plot(x,y, color='red') 
plt.show()
```
